5.Напишите функцию, которая принимает список чисел и возвращает их среднее арифметическое.


```python
my_list = [int(elem) for elem in input().split()]   # ввод элементов с клавиатуры через ' '
average=sum(my_list)/len(my_list)     # нахождение ср.арифметического
print(average)
```
