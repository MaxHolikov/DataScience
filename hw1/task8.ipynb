8.Создайте словарь с пятью парами «ключ-значение», где ключи - названия фруктов, а значения - их цвета. Выведите только те фрукты, цвет которых содержит букву 'к'.



```python
fruits = {
    "Яблоко": "красный",
    "Лимон": "желтый",
    "Банан": "желтый",
    "Апельсин": "оранжевый",
    "Вишня": "темно-красный"
}

for fruit, color in fruits.items():     # Вывод фруктов, цвет которых содержит букву 'к' (игнорируя регистр)
    if 'к' in color.lower():
        print(f"{fruit}: {color}")
```
