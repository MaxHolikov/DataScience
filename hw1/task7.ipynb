7.Напишите программу, которая принимает от пользователя целое число и выводит таблицу умножения для этого числа от 1 до 10.



```python
try:
    num = int(input("Введите целое число: "))   # Запрос числа у пользователя с клавиатуры
    print(f"Таблица умножения для числа {num}:")
    for i in range(1, 11):
      print(f"{num} x {i} = {num * i}")

except ValueError:        # работа с некорректным вводом числа
    print("Пожалуйста, введите корректное целое число.")
```
