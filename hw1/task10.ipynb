10. Создайте список из 5 строк и замените в нем каждый третий символ в каждой строке на звездочку '*'.


```python
my_list = ['один', 'два', 'три', 'четыре', 'пять']
new_list=[]
for elem in my_list:
  chars = list(elem)
  for i in (range(len(my_list))):
    if i==2:
      chars[i]='*'
  new_list=''.join(chars)
  print(new_list)
```
