9.Напишите функцию, которая принимает строку и возвращает True, если строка является палиндромом (игнорируя регистр и пробелы), и False в противном случае.

```python
text = str(input())                     # воод текста с клавиатуры
text = text.lower().replace(' ', '')    # приведение к нижнему регистру и удаление пробеллов
a = text[::-1]
if text == a:
   print ('True')
else:
     print ('False')
```
